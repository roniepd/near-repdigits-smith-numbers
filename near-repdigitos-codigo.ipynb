```python
from sympy import factorint
def sd(n):
    return sum(map(int, str(n)))
def is_smith(n):
    f = factorint(n)
    return sum(f.values()) > 1 and \
           sd(n) == sum(e*sd(p) for p, e in f.items())
def near_repdigits(m):
    A = set()
    for a in range(10):
        for b in range(10):
            if a == b:
                continue
            for j in range(m):
                d = [str(a)]*m
                d[m-1-j] = str(b)
                if d[0] != '0':
                    A.add(int(''.join(d)))
    return sorted(A)
A = []
for m in range(3, 20):
    A += [n for n in near_repdigits(m) if is_smith(n)]
print(A)
```
